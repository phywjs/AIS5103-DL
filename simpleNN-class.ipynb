```python
import torch
import torch.nn as nn
class SimpleNN(nn.Module):
  def __init__(self):
    super().__init__()
    self.hidden = nn.Linear(2,4)
    self.output = nn.Linear(4,1)

  def forward(self,x):
    h = torch.relu(self.hidden(x))
    x = self.output(h)
    return x

model = SimpleNN()
x = torch.tensor([[1.0,2.0]])
y = model(x)
print("y=", y)
```

```python
x2 = torch.tensor([[1.0,2.0],[2.0,4.0]])
```

```python
y2 = model(x2)
print(y2)
```
